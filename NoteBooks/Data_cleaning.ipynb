```python
import pandas as pd
import numpy as np

df = pd.read_csv(r"C:\Users\mshlo\Downloads\diabetes+130-us+hospitals+for+years+1999-2008\diabetic_data.csv")

print("Raw shape:", df.shape)
df = df.replace("?", np.nan)
df["target"] = (df["readmitted"] == "<30").astype(int)
encounter_ids = df["encounter_id"].copy()

missing_summary = (
    df.isnull()
      .mean()
      .mul(100)
      .sort_values(ascending=False)
)

print("\nMissing percentage:")
print(missing_summary.head(15))

df = df.drop(
    columns=["weight", "payer_code", "medical_specialty"]
)

df = df.drop(
    columns=["encounter_id", "patient_nbr"]
)
df = df.drop(
    columns=["readmitted"]
)

print("\nFinal shape:", df.shape)

print("\nRemaining columns:")
print(df.columns.tolist())

print("\nRemaining missing values:")
print(df.isnull().sum().sort_values(ascending=False).head(15))

print("\nTarget distribution:")
print(df["target"].value_counts())

print("\nTarget percentage:")
print(df["target"].value_counts(normalize=True) * 100)

df.to_csv(r"C:\Users\mshlo\Downloads\diabetes+130-us+hospitals+for+years+1999-2008\CleanedDiabetic_data.csv", index=False)

print("cleaned.csv saved successfully.")
```
